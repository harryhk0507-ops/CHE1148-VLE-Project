```python
import pandas as pd
import re
from itables import show

df1 = pd.read_csv("binary_5plus_high_row_count_part1.csv")
df2 = pd.read_csv("binary_5plus_high_row_count_part2.csv")
# Combine with df2 below df1
df = pd.concat([df1, df2], axis=0, ignore_index=True)
```

```python
# Remove columns where ALL values are NaN/empty
df = df.dropna(axis=1, how="all")
```

```python
print(df.shape)
print(df.iloc[1])
show(df.iloc[1])
```

```python
print(df.columns.tolist())
```

```python
print(df["phase"].unique())
```

```python
df_phase = df[df["phase"].isin(["Gas", "Liquid"])]
print(f"Shape after dropping: {df_phase.shape}")
```

```python
print(df_phase["property"].unique())
```

```python
# Drop rows where Temperature,K is NaN
df_phase_T = df_phase.dropna(subset=["Temperature, K"], how="all")

print(f"Shape after dropping: {df_phase_T.shape}")
```

```python
print(df_phase_T["property"].unique())
```

```python
# Drop rows where ALL specified columns about Pressure (P) are NaN
df_phase_T_P = df_phase_T.dropna(
    subset=[
        "Pressure, kPa",
        "Pressure, kPa_1",
        "Pressure, kPa_2"
    ],
    how="all"
)

print(f"Shape after dropping: {df_phase_T_P.shape}")
```

```python
# Remove columns where ALL values are NaN/empty
df_phase_T_P_columns = df_phase_T_P.dropna(axis=1, how="all")

print(f"Shape after dropping: {df_phase_T_P_columns.shape}")
print(df_phase_T_P_columns.columns.tolist())
```

```python
# Drop rows where ALL specified columns about mole fractions are NaN
df_phase_T_P_columns.loc[
    df_phase_T_P_columns["property"] == "mole fraction",
    "Mole fraction"
] = df_phase_T_P_columns["value"]
df_phase_T_P_mf = df_phase_T_P_columns.dropna(subset=["Mole fraction"], how="all")

print(f"Shape after dropping: {df_phase_T_P_mf.shape}")
```

```python
df_phase_T_P_mf.to_csv("Halfway_Cleaned_VLE_Data.csv")
```

```python
# Replace 'Column1', 'Column2', etc. with the actual names of the columns you want to keep
df_clean = df_phase_T_P_mf[[
    "property",
    "value",
    "phase",
    "Temperature, K",
    "Pressure, kPa",
    "Mole fraction",
    "mix_key"
]]
```

```python
show(df_clean.head())
```

```python
df_clean[["Component 1", "Component 2"]] = (
    df_clean["mix_key"]
    .str.split("|", expand=True, regex=False)
)
df_clean.drop(columns=["mix_key"], inplace=True)
show(df_clean.head())
```

```python
df_clean.to_csv("Cleaned_VLE_Data.csv", index=False)
```
